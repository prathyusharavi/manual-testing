```python
ids = [1, 2, 3, 1, 4, 5]
seen = set()
left = 0
max_length = 0
for right in range(len(ids)):
    while ids[right] in seen:
        seen.remove(ids[left])
        left += 1
    seen.add(ids[right])
    length = right - left + 1
    max_length = max(max_length, length)
print(max_length)
```

```python
discounts = [-2, 3, 4, -1, 5, -6]
current_sum = discounts[0]
max_sum = discounts[0]
for i in range(1, len(discounts)):
    current_sum = max(discounts[i], current_sum + discounts[i])
    max_sum = max(max_sum, current_sum)
print(max_sum)
```

```python
height = [0, 1, 0, 2, 1, 0, 1, 3]
water = 0
for i in range(1, len(height) - 1):
    left_max = max(height[:i])
    right_max = max(height[i+1:])
    water += max(0, min(left_max, right_max) - height[i])
```

```python
numbers = [2, 3, -2, 4]
current_max = numbers[0]
current_min = numbers[0]
answer = numbers[0]
for i in range(1, len(numbers)):
    num = numbers[i]
    old_max = current_max
    old_min = current_min
    current_max = max(num, old_max * num, old_min * num)
    current_min = min(num, old_max * num, old_min * num)
    answer = max(answer, current_max)
print(answer)
```

```python
skills = ["eat", "tea", "tan", "ate", "nat", "bat"]
groups = {}
for word in skills:
    key = ''.join(sorted(word))
    if key not in groups:
        groups[key] = []
    groups[key].append(word)
print(list(groups.values()))
```

```python
numbers = [100, 4, 200, 1, 3, 2]
numbers.sort()
longest = 1
current = 1
for i in range(1, len(numbers)):
    if numbers[i] == numbers[i - 1] + 1:
        current += 1
    else:
        current = 1
    longest = max(longest, current)
print(longest)
```

```python
intervals = [[1,3], [2,6], [8,10], [9,12]]
intervals.sort()
merged = []
for interval in intervals:
    if not merged or merged[-1][1] < interval[0]:
        merged.append(interval)
    else:
        merged[-1][1] = max(merged[-1][1], interval[1])
print(merged)
```
